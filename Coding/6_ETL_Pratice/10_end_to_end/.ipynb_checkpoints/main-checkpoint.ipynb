```python
import pandas as pd

folder_path = "./Dataset"
```

```python
customers = pd.read_csv(folder_path + "/customers.csv")
orders = pd.read_csv(folder_path + "/orders.csv")
products = pd.read_csv(folder_path + "/products.csv")
shippings = pd.read_csv(folder_path + "/shipping.csv")
payments = pd.read_json(folder_path + "/payment.json")
```

```python
dfs = {
    "Customers" : customers,
    "Orders" : orders,
    "Products" : products,
    "Shippings" : shippings,
    "Payments" : payments
}
```

```python
for key, value in dfs.items():
    print("DataFrame: ", key)
    print("Columns: ", ", ".join(value.columns))
    print()
```

```python
for key, value in dfs.items():
    print("=" * 40)
    print("DataFrame:", key)

    print("\nMissing Values:")
    print(value.isnull().sum())

    print("\nDuplicate Records:")
    print(value.duplicated().sum())

    print("\nData Types:")
    print(value.dtypes)

    print()
```

```python
invalid_customers = orders[
    ~orders["CustomerID"].isin(customers["CustomerID"])
]

print("Invalid Customer IDs:")
print(invalid_customers)
```

```python
customers["CustomerName"] = customers["CustomerName"].fillna("Unknown")

orders = orders.drop_duplicates()
```

```python
orders.drop(index=8, inplace=True)
```

```python
category_price_avg = products.groupby("Category")["Price"].mean()
```

```python
def fill_price(row):
    if pd.isna(row["Price"]):
        row["Price"] = round(category_price_avg.loc[row["Category"]], 2)
    return row

products = products.apply(fill_price, axis= 1)
```

```python
products
```

```python
sales = orders.merge(
    customers,
    on="CustomerID",
    how="left"
)


sales = sales.merge(
    products,
    on="ProductID",
    how="left"
)


sales = sales.merge(
    payments,
    on="OrderID",
    how="left"
)


sales = sales.merge(
    shippings,
    on="OrderID",
    how="left"
)
```

```python
sales
```

```python
sales["TotalSales"] = sales["Quantity"] * sales["Price"]
```

```python
total_sales = sales["TotalSales"].sum()
total_orders = sales["OrderID"].nunique()
top_products = sales.groupby("ProductName")["Quantity"].sum().sort_values(ascending=False).head(10)
sales_by_city = sales.groupby("City")["TotalSales"]
    .sum()
    .reset_index()
)
```
