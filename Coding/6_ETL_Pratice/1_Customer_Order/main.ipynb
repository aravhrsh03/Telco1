```python
import pandas as pd

ordersdf = pd.read_csv("./Dataset/orders.csv")
customersdf = pd.read_csv("./Dataset/customers.csv")
productsdf = pd.read_csv("./Dataset/products.csv")
```

```python
ordersdf.head()
```

```python
productsdf.head()
```

```python
dfs = {
    "Order": ordersdf,
    "Customer": customersdf,
    "Product": productsdf
}

for key, value in dfs.items():
    print("DataFrame: ", key)
    print("Columns: ", ", ".join(value.columns.tolist()))
    print()
```

```python
# Missing customer names

dfs["Customer"]["CustomerName"] = dfs["Customer"]["CustomerName"].fillna("Unknown")
```

```python
dfs["Customer"]
```

```python
orders_customer_df = dfs["Customer"].merge(dfs["Order"], on="CustomerID")
```

```python
orders_customer_df
```

```python
orders_customers_products_df = orders_customer_df.merge(dfs["Product"], on = "ProductID")
```

```python
orders_customers_products_df
```

```python
orders_customers_products_df["TotalSales"] = orders_customers_products_df["Quantity"] * orders_customers_products_df["Price"]
```

```python
orders_customers_products_df
```

```python
final_sales_df = orders_customers_products_df[["OrderID", "CustomerName", "Product", "TotalSales"]]
```

```python
final_sales_df
```

```python
final_sales_df.columns = ["OrderID", "Customer", "Product", "TotalSales"]
```

```python
final_sales_df
```

```python
final_sales_df.to_csv("")
```
