```python
import pandas as pd

employee_df = pd.read_csv("./Dataset/employee.csv")
```

```python
employee_df
```

```python
department_avg = employee_df.groupby("Department")["Salary"].mean()
```

```python
department_avg
```

```python
def fill_salary(row):
    if pd.isna(row["Salary"]):
        dept = row["Department"]
        match dept:
            case "HR":
                row["Salary"] = department_avg["HR"]
            case "IT":
                row["Salary"] = department_avg["IT"]
            case "Sales":
                row["Salary"] = department_avg["Sales"]
    return row
                
```

```python
employee_df = employee_df.apply(fill_salary, axis = 1)
```

```python
employee_df
```

```python
employee_df["Bonus"] = employee_df["Salary"] * 1.10
```

```python
employee_df
```

```python
final_employee_df = employee_df[["Name", "Salary", "Bonus"]]
```
