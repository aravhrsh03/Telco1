```python
import pandas as pd

bank_df = pd.read_csv("./Dataset/bank.csv")
```

```python
bank_df
```

```python
bank_df["Balance"] = bank_df["Deposit"] - bank_df["Withdraw"]
```

```python
bank_df
```

```python
def account_status(row):
    status = False
    if row["Balance"] >= 0:
        status = True

    row["Status"] = "OK" if status else "Overdrawn"
    return row
```

```python
bank_df = bank_df.apply(account_status, axis = 1)
```
