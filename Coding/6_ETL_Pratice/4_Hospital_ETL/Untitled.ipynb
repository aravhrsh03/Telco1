```python
import pandas as pd

patient_df = pd.read_csv("./Dataset/patient.csv")
```

```python
patient_df
```

```python
avg_age = patient_df.loc[
    (patient_df["Age"] < 120) & (patient_df["Age"].notna()),
    "Age"
].mean()
```

```python
patient_df.loc[
    (patient_df["Age"] > 120) & (patient_df["Age"].isna()),
    "Age"
] = avg_age
```

```python
patient_df
```

```python
patient_df = patient_df.fillna("Unknown")
```
