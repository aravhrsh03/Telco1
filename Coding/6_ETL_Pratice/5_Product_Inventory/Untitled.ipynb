```python
import pandas as pd

inventory_df = pd.read_csv("./Dataset/inventory.csv")
```

```python
inventory_df
```

```python
inventory_df.loc[inventory_df["Stock"] < 0, "Stock"] = 0
```

```python
inventory_df
```

```python
inventory_df.loc[inventory_df["Stock"].isna(), "Stock"] = 0
```

```python
inventory_df
```

```python
def fill_status(stock):
    if stock == 0:
        return "Out of Stock"
    elif stock >= 1 and stock < 11:
        return "Low"
    else:
        return "Available"
```

```python
inventory_df["Status"] = inventory_df["Stock"].apply(fill_status)
```
