```python
import pandas as pd

student_df = pd.read_csv("./Dataset/student.csv")
```

```python
student_df
```

```python
student_df["Science"] = student_df["Science"].fillna(student_df["Science"].mean())
```

```python
student_df
```

```python
student_df["Total"] = student_df["Math"] + student_df["Science"]
```

```python
student_df
```

```python
student_df["Percentage"] = student_df["Total"] / 2
```

```python
student_df
```

```python
def fill_grade(percent):
    if 90 <= percent <= 100:
        return "A"
    elif 75 <= percent <= 89:
        return "B"
    elif 50 <= percent <= 74:
        return "C"
    elif percent < 50:
        return "D"
    else:
        return "Unknown"
```

```python
student_df["Grade"] = student_df["Percentage"].apply(fill_grade)
```

```python
student_df
```

```python
final_student_df = student_df[["Student", "Grade"]]
```

```python
final_student_df
```

```python
final_student_df.to_csv("./final_student.csv")
```
